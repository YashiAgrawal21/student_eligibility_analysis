```python
import pandas as pd
```

```python
data = {
    "student_id": [101, 102, 103, 104, 105, 106],
    "attendance_percent": [92, 67, 81, 45, 74, 88],
    "assignment_score": [18, 12, 15, 8, 14, 19],
    "quiz_score": [72, 48, 65, 30, 55, 80],
    "lab_completed": [True, False, True, False, True, True]
}

df = pd.DataFrame(data)

df
```

```python
df.describe()
```

```python
df['total_score'] = df['assignment_score'] + df['quiz_score']
df
```

```python
df['eligible'] = (
    (df['attendance_percent'] >= 75) &
    (df['total_score'] >= 70) &
    (df['lab_completed'])
)
df
```

```python
eligible_students = df[df["eligible"] == True]
eligible_students
```

```python
# Total score
df["total_score"] = df["assignment_score"] + df["quiz_score"]

# Average score
df["average_score"] = df["total_score"] / 2

# Performance category
df["performance"] = pd.cut(
    df["total_score"],
    bins=[0, 60, 80, 100],
    labels=["Low", "Medium", "High"]
)

# Attendance category
df["attendance_category"] = pd.cut(
    df["attendance_percent"],
    bins=[0, 60, 75, 100],
    labels=["Poor", "Average", "Good"]
)

print(df.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.bar(df["student_id"].astype(str), df["total_score"])
plt.title("Total Score by Student")
plt.xlabel("Student ID")
plt.ylabel("Total Score")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["attendance_percent"], df["total_score"])
plt.title("Attendance vs Total Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Total Score")
plt.show()
```

```python
eligibility_counts = df["eligible"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    eligibility_counts,
    labels=["Eligible", "Not Eligible"],
    autopct="%1.1f%%"
)
plt.title("Eligibility Distribution")
plt.show()
```

```python
performance_counts = df["performance"].value_counts()

plt.figure(figsize=(6,4))
performance_counts.plot(kind="bar")
plt.title("Performance Categories")
plt.xlabel("Category")
plt.ylabel("Number of Students")
plt.show()
```
